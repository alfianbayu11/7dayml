```python


import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import warnings

nltk.download('punkt')
warnings.filterwarnings("ignore")

# Sample E-commerce FAQ data (replace with your actual data)
faq_data = {
    "questions": [
        "How do I place an order?",
        "What are your shipping costs?",
        "What payment methods do you accept?",
        "How long does shipping take?",
        "What is your return policy?",
        "Do you offer international shipping?",
        "How do I track my order?",
        "Can I cancel my order?",
        "What if my order arrives damaged?",
        "How do I contact customer support?"
    ],
    "answers": [
        "To place an order, browse our products, add them to your cart, and proceed to checkout.",
        "Shipping costs vary depending on your location and the weight of your order. They are calculated at checkout.",
        "We accept Visa, Mastercard, American Express, and PayPal.",
        "Shipping times depend on your location, typically between 3-7 business days.",
        "We offer a 30-day return policy on most items. See our website for details.",
        "Yes, we offer international shipping to select countries.",
        "You can track your order using the tracking number provided in your shipping confirmation email.",
        "You can cancel your order within 24 hours of placement. Contact customer support for assistance.",
        "If your order arrives damaged, please contact customer support immediately.",
        "You can contact customer support via email at support@example.com or by phone at 1-800-123-4567."
    ]
}


# Create a TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Fit and transform the questions to create a TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(faq_data["questions"])

def get_answer(question):
    # Transform the user's question into a TF-IDF vector
    user_question_vector = vectorizer.transform([question])

    # Calculate cosine similarity between the user's question and all FAQs
    similarities = cosine_similarity(user_question_vector, tfidf_matrix)

    match_index=similarities.argmax()
    match_score=similarities[0,match_index]
    threshold=0.1
    if match_score>=threshold:
      return faq_data["answers"][match_index]
    else:
      return "I'm sorry, I don't have an answer for that question."
```

```python
while True:
    user_input = input("User: ")
    if user_input.lower() == 'exit':
        print("Chatbot: Goodbye!")
        break
    response = get_answer(user_input)
    print("Chatbot:", response)
```
